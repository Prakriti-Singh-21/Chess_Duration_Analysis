```python
#outlier detection

from scipy import stats

from scipy import stats
import pandas as pd

# Numerical columns for the z-score 
col = ['turns', 'white_rating', 'black_rating', 'opening_ply']

# Calculating z-scores 
z_scores = stats.zscore(chess[col])

# Z-score greater than 3
outlier = (abs(z_scores) > 3).sum(axis=0)

# Summary table 
outlier_table = pd.DataFrame({
    'Column Name': col,
    'Outliers (|Z| > 3)': outlier
})

# Display the summary table
display(outlier_table)
```

```python
#Plotting summary table

#Visual style
sns.set_theme(style= 'whitegrid')

plt.figure(figsize= (8,5))

#Plot
bar = sns.barplot( data = outlier_table, x='Column Name', y =  'Outliers (|Z| > 3)',palette = 'viridis',hue='Column Name',legend = False)
plt.title = ('Distribution of Outliers(|Z| > 3)')
plt.xlabel('Columns')
plt.ylabel('Outlier Count')            

plt.show()

    
```

```python
%%writefile app.py
import streamlit as st
import pandas as pd
import altair as alt

st.set_page_config(page_title="Chess Opening Explorer", layout="wide")
st.title("♟️ Chess Opening & Duration Explorer")
st.write("Explore the relationship between chess opening families, player skill tiers, and game duration.")

@st.cache_data
def load_data():
    df = pd.read_csv(r"C:\Users\Prakriti\Desktop\MSU Study\Fall 2026\CMSE 830\Home work\games_cleaned.csv")
    df['avg_rating'] = (df['white_rating'] + df['black_rating']) / 2
    return df

try:
    chess = load_data()
except Exception as e:
    st.error(f"Error loading dataset: {e}")
    st.stop()

def get_skill_tier(rating):
    if rating < 1400:
        return 'Beginner (<1400)'
    elif rating < 1800:
        return 'Intermediate (1400-1800)'
    else:
        return 'Advanced / Pro (1800+)'

chess['skill_tier'] = chess['avg_rating'].apply(get_skill_tier)

st.sidebar.header("Filter Controls")
top_families = chess['opening_family'].value_counts().head(8).index.tolist()
selected_families = st.sidebar.multiselect(
    "Select Opening Families:",
    options=top_families,
    default=top_families[:4]
)

filtered_df = chess[chess['opening_family'].isin(selected_families)]

st.metric("Total Games in View", f"{len(filtered_df):,}")

if len(filtered_df) > 0:
    bar_stats = filtered_df.groupby(['opening_family', 'skill_tier']).agg(
        mean_turns=('turns', 'mean'),
        game_count=('turns', 'count')
    ).reset_index()

    # Using Altair chart as styled in your coursework template
    chart = alt.Chart(bar_stats).mark_bar().encode(
        x=alt.X('opening_family:N', title='Opening Family'),
        y=alt.Y('mean_turns:Q', title='Mean Turns'),
        color=alt.Color('skill_tier:N', title='Skill Tier'),
        column='skill_tier:N'
    ).properties(
        title="Average Game Duration by Opening Family & Skill Tier",
        width=200,
        height=400
    )
    
    st.altair_chart(chart, use_container_width=True)
else:
    st.warning("Please select at least one opening family.")
```

```python
import plotly.graph_objects as go
import pandas as pd
import scipy.stats as stats

# 1. Filter for top 8 opening families
top_families = chess['opening_family'].value_counts().head(8).index
filtered_chess = chess[chess['opening_family'].isin(top_families)]

# 2. Group by opening family to calculate average ply, turns, and game counts
family_rel = filtered_chess.groupby('opening_family').agg(
    mean_ply=('opening_ply', 'mean'),
    mean_turns=('turns', 'mean'),
    game_count=('turns', 'count')
).reset_index()

# Normalize bubble sizes to a clean visual range (12 to 45 pixels)
min_cnt = family_rel['game_count'].min()
max_cnt = family_rel['game_count'].max()
family_rel['bubble_size'] = 12 + (family_rel['game_count'] - min_cnt) / (max_cnt - min_cnt + 1e-5) * 33

# 3. Overall statistics and correlation
overall_mean_turns = filtered_chess['turns'].mean()
total_games = len(filtered_chess)
corr_coef, p_val = stats.pearsonr(filtered_chess['opening_ply'], filtered_chess['turns'])

# 4. Create Figure using Graph Objects (guarantees unique colors & correct sizing)
fig = go.Figure()

colors = [
    '#636EFA', '#EF553B', '#00CC96', '#AB63FA', 
    '#FFA15A', '#19D3F3', '#FF6692', '#B6E880'
]

for i, row in family_rel.iterrows():
    fig.add_trace(go.Scatter(
        x=[row['mean_ply']],
        y=[row['mean_turns']],
        mode='markers+text',
        name=row['opening_family'],
        text=row['opening_family'],
        textposition='top center',
        marker=dict(
            size=row['bubble_size'],
            color=colors[i % len(colors)],
            opacity=0.85,
            line=dict(width=1, color='DarkSlateGrey')
        ),
        hovertemplate=(
            f"<b>{row['opening_family']}</b><br>"
            f"Average Opening Depth (Ply): {row['mean_ply']:.2f}<br>"
            f"Average Total Game Duration (Turns): {row['mean_turns']:.2f}<br>"
            f"Total Games Played: {row['game_count']:,}<extra></extra>"
        ),
        showlegend=False
    ))

# 5. Dataset-wide baseline reference line for average game length
fig.add_hline(
    y=overall_mean_turns,
    line_dash="dash",
    line_color="gray",
    annotation_text=f"Overall Mean Turns ({overall_mean_turns:.1f})",
    annotation_position="bottom right"
)

# 6. Quantitative Summary Box 
stats_text = (
    f"<b>Quantitative Summary</b><br>"
    f"• Overall Mean: <b>{overall_mean_turns:.1f} turns</b><br>"
    f"• Pearson r: <b>{corr_coef:.3f}</b><br>"
    f"• p-value: <b>< 0.001</b><br>"
    f"• Sample Size (n): <b>{total_games:,}</b>"
)

fig.add_annotation(
    x=5.2, 
    y=59.0,
    xref="x", 
    yref="y",
    text=stats_text,
    showarrow=True,
    arrowhead=2,
    arrowsize=1,
    arrowwidth=1,
    arrowcolor="#666666",
    ax=40,
    ay=-40,
    font=dict(size=12, color="#1e293b"),
    align="left",
    bgcolor="rgba(255, 255, 255, 0.92)",
    bordercolor="#cbd5e1",
    borderwidth=1,
    borderpad=8
)

# 7. Layout adjustments 
fig.update_layout(
    title='<b>Relating Opening Depth, Game Duration, and Popularity Across Top 8 Families</b>',
    xaxis_title='Average Opening Depth (Ply)',
    yaxis_title='Average Total Game Duration (Turns)',
    width=950,
    height=600,
    font=dict(size=13, family="Arial"),
    title_font=dict(size=16),
    margin=dict(l=60, r=40, t=80, b=60)
)

fig.show()
```

```python
import plotly.graph_objects as go
import pandas as pd

# 1. Filter for top 8 opening families and calculating mean and sample size
top_families = chess['opening_family'].value_counts().head(8).index
filtered_chess = chess[chess['opening_family'].isin(top_families)]

family_stats = filtered_chess.groupby('opening_family')['turns'].agg(
    mean_turns='mean',
    median_turns='median',
    game_count='count'
).reset_index()

# Sort from shortest to longest average duration 
df_sorted = family_stats.sort_values(by='mean_turns', ascending=True)

# 2.Quantitative metrics 
overall_mean = filtered_chess['turns'].mean()
total_games = len(filtered_chess)
longest_family = df_sorted.iloc[-1]['opening_family']
longest_val = df_sorted.iloc[-1]['mean_turns']
shortest_family = df_sorted.iloc[0]['opening_family']
shortest_val = df_sorted.iloc[0]['mean_turns']

# 3. Line Chart 
fig = go.Figure()

fig.add_trace(go.Scatter(
    x=df_sorted['mean_turns'],
    y=df_sorted['opening_family'],
    mode='lines+markers',
    marker=dict(size=12, color='royalblue', line=dict(width=1, color='darkblue')),
    line=dict(color='royalblue', width=3, dash='solid'),
    hovertemplate='<b>%{y}</b><br>Average Turns: %{x:.1f}<extra></extra>'
))

# 4.Vertical reference line 
fig.add_vline(
    x=overall_mean, 
    line_dash="dash", 
    line_color="crimson", 
    annotation_text=f"Dataset Mean: {overall_mean:.1f}", 
    annotation_position="top"
)

# 5. Quantitative summary c
stats_text = (
    f"<b>Quantitative Summary</b> | "
    f"Overall Mean: <b>{overall_mean:.1f}</b> | "
    f"Longest ({longest_family}): <b>{longest_val:.1f}</b> | "
    f"Shortest ({shortest_family}): <b>{shortest_val:.1f}</b> | "
    f"n = <b>{total_games:,}</b>"
)

fig.add_annotation(
    x=0.5, 
    y=0.98,
    xref="paper", 
    yref="paper",
    text=stats_text,
    showarrow=False,
    font=dict(size=11, color="black"),
    align="center",
    bgcolor="rgba(255, 255, 255, 0.95)",
    bordercolor="#333333",
    borderwidth=1,
    borderpad=6
)

# 6. Final layout styling
fig.update_layout(
    title='<b>Top eight Opening Families with shortest vs longest game duration</b>',
    xaxis_title='Average Total Turns',
    yaxis_title='Opening Family',
    width=950,
    height=550,
    font=dict(size=13, family="Arial"),
    margin=dict(l=160, r=40, t=80, b=60)
)

fig.show()
```
