## 1.3. Data integrity and outlier assessment

Following are the steps to conduct the assessment:

1. Duplicate Value Check `chess.duplicated().sum()`: It Scans the dataset for identical, redundant rows to prevent skewness.
2. Missing Value Audit `chess.isnull().sum()):` Counts the number of NaN or null values per column to determine if data imputation or row dropping is necessary.
3. Extreme Outlier Detection `stats.zscore with $\vert{}Z\vert{} > 3$`: This uses the Z-scores to identify extreme values that deviate more than 3 standard deviations from the mean in numerical columns like turns, ratings, and opening_ply.

```python
from scipy import stats

# Duplicate value check
duplicate_count = chess.duplicated().sum()
print(f"Number of duplicate rows: {duplicate_count}")


#Missing value
missing =chess.isnull().sum()
print(f'Number of missing values: {missing}')


#extreme outlier - |Z score| > 3
col = ['turns','white_rating', 'black_rating', 'opening_ply']
z_scores = stats.zscore(chess[col])
outlier = (abs(z_scores) > 3).sum(axis=0)
# Summary table 
outlier_table = pd.DataFrame({
    'Column Name': col,
    'Outliers (|Z| > 3)': outlier
})

# Display the summary table for the outliers detected
display(outlier_table)
```

## 1.4 Data cleaning and Preparation

    - Numerical Synthesis: sums `white_rating` and `black_rating` for each game and divides by 2 constructs a single representative metric `(avg_rating)` to quantify the overall skill level of a match
    - Binary Indicator Encoding : Encodes  a continuosus varaible to a  Binary variable by converting it into into 1 for adavance players and 0 for intermidiate and begginer level players

## 2.1.1. Line Plot:

**Details of the Code:**

1. **Data Aggregation:** Filters the dataset for the top 8 frequent opening families and calculates their average (`mean_turns`), median, and total game count, sorting them from shortest to longest duration.
2. **Metric Extraction:** Computes overall benchmark values (the global dataset mean, total sample size, and identifies the absolute shortest/longest opening families).
3. **Chart Construction:** Plots a horizontal line-and-marker chart where each point represents an opening family's average game length.
4. **Reference & Annotations:** Adds a **crimson dashed vertical line** representing the overall dataset mean and injects a formatted **summary box** at the top with key quantitative metrics ($n$, overall mean, extremes).
5. **Layout Styling:** Configures custom dimensions, titles, axis labels, and clean margins for presentation-ready visualization.



```python
#Visulisation :

import plotly.graph_objects as go
import pandas as pd

# 1. Filter for top 8 opening families and calculating mean and sample size
top_families = chess['opening_family'].value_counts().head(8).index
filtered_chess = chess[chess['opening_family'].isin(top_families)]

family_stats = filtered_chess.groupby('opening_family')['turns'].agg(
    mean_turns='mean',
    median_turns='median',
    game_count='count'
).reset_index()

# Sort from shortest to longest average duration 
df_sorted = family_stats.sort_values(by='mean_turns', ascending=True)

# 2.Quantitative metrics 
overall_mean = filtered_chess['turns'].mean()
total_games = len(filtered_chess)
longest_family = df_sorted.iloc[-1]['opening_family']
longest_val = df_sorted.iloc[-1]['mean_turns']
shortest_family = df_sorted.iloc[0]['opening_family']
shortest_val = df_sorted.iloc[0]['mean_turns']

# 3. Line Chart 
fig = go.Figure()

fig.add_trace(go.Scatter(
    x=df_sorted['mean_turns'],
    y=df_sorted['opening_family'],
    mode='lines+markers',
    marker=dict(size=12, color='royalblue', line=dict(width=1, color='darkblue')),
    line=dict(color='royalblue', width=3, dash='solid'),
    hovertemplate='<b>%{y}</b><br>Average Turns: %{x:.1f}<extra></extra>'
))

# 4.Vertical reference line 
fig.add_vline(
    x=overall_mean, 
    line_dash="dash", 
    line_color="crimson", 
    annotation_text=f"Dataset Mean: {overall_mean:.1f}", 
    annotation_position="top"
)

# 5. Quantitative summary c
stats_text = (
    f"<b>Quantitative Summary</b> | "
    f"Overall Mean: <b>{overall_mean:.1f}</b> | "
    f"Longest ({longest_family}): <b>{longest_val:.1f}</b> | "
    f"Shortest ({shortest_family}): <b>{shortest_val:.1f}</b> | "
    f"n = <b>{total_games:,}</b>"
)

fig.add_annotation(
    x=0.5, 
    y=0.98,
    xref="paper", 
    yref="paper",
    text=stats_text,
    showarrow=False,
    font=dict(size=11, color="black"),
    align="center",
    bgcolor="rgba(255, 255, 255, 0.95)",
    bordercolor="#333333",
    borderwidth=1,
    borderpad=6
)

# 6. Final layout styling
fig.update_layout(
    title='<b>Top eight Opening Families with shortest vs longest game duration</b>',
    xaxis_title='Average Total Turns',
    yaxis_title='Opening Family',
    width=950,
    height=550,
    font=dict(size=13, family="Arial"),
    margin=dict(l=160, r=40, t=80, b=60)
)

fig.show()
```

## 2.1.2. Bubble Chart

**Details of the code**

This codes builds an interactive bubble chart to analyze the relationship between three variables across the top 8 chess opening families:

1. **Average Opening Depth (Ply)** on the X-axis.
2. **Average Total Game Duration (Turns)** on the Y-axis.
3. **Game Popularity / Volume of an opening family** represented by bubble size.

Following are the steps performed:

1. **Data Filtering & Aggregation:** Isolates the top 8 opening families and calculates their average opening ply, average game turns, and total game counts. It also scales the game counts to determine the bubble sizes.
2. **Statistical Calculation:** Computes the dataset-wide mean turns and runs a **Pearson correlation test** (`stats.pearsonr`) between opening depth and game duration.
3. **Chart Construction:** Generates custom scatter bubbles for each opening family, complete with text labels above the markers and detailed hover tooltips.
4. **Reference Line & Annotation Box:** Adds a horizontal dashed line for the overall mean turns baseline and injects a formatted Quantitative
Summary box pointing to key metrics (sample size, overall mean, and Pearson $r$).
6. **Layout Styling:** Sets custom dimensions, clean fonts, titles, and margins for a polished visualization.

```python
import plotly.graph_objects as go
import pandas as pd
import scipy.stats as stats

# 1. Filter for top 8 opening families
top_families = chess['opening_family'].value_counts().head(8).index
filtered_chess = chess[chess['opening_family'].isin(top_families)]

# 2. Group by opening family 
family_rel = filtered_chess.groupby('opening_family').agg(
    mean_ply=('opening_ply', 'mean'),
    mean_turns=('turns', 'mean'),
    game_count=('turns', 'count')
).reset_index()

# Normalize bubble sizes 
min_cnt = family_rel['game_count'].min()
max_cnt = family_rel['game_count'].max()
family_rel['bubble_size'] = 12 + (family_rel['game_count'] - min_cnt) / (max_cnt - min_cnt + 1e-5) * 33

# 3. Overall statistics and correlation
overall_mean_turns = filtered_chess['turns'].mean()
total_games = len(filtered_chess)
corr_coef, p_val = stats.pearsonr(filtered_chess['opening_ply'], filtered_chess['turns'])

# 4. Create Figure using Graph Objects (guarantees unique colors & correct sizing)
fig = go.Figure()

colors = [
    '#636EFA', '#EF553B', '#00CC96', '#AB63FA', 
    '#FFA15A', '#19D3F3', '#FF6692', '#B6E880'
]

for i, row in family_rel.iterrows():
    fig.add_trace(go.Scatter(
        x=[row['mean_ply']],
        y=[row['mean_turns']],
        mode='markers+text',
        name=row['opening_family'],
        text=row['opening_family'],
        textposition='top center',
        marker=dict(
            size=row['bubble_size'],
            color=colors[i % len(colors)],
            opacity=0.85,
            line=dict(width=1, color='DarkSlateGrey')
        ),
        hovertemplate=(
            f"<b>{row['opening_family']}</b><br>"
            f"Average Opening Depth (Ply): {row['mean_ply']:.2f}<br>"
            f"Average Total Game Duration (Turns): {row['mean_turns']:.2f}<br>"
            f"Total Games Played: {row['game_count']:,}<extra></extra>"
        ),
        showlegend=False
    ))

# 5. Dataset-wide baseline
fig.add_hline(
    y=overall_mean_turns,
    line_dash="dash",
    line_color="gray",
    annotation_text=f"Overall Mean Turns ({overall_mean_turns:.1f})",
    annotation_position="bottom right"
)

# 6. Quantitative Summary Box 
stats_text = (
    f"<b>Quantitative Summary</b><br>"
    f"• Overall Mean: <b>{overall_mean_turns:.1f} turns</b><br>"
    f"• Pearson r: <b>{corr_coef:.3f}</b><br>"
    f"• p-value: <b>< 0.001</b><br>"
    f"• Sample Size (n): <b>{total_games:,}</b>"
)

fig.add_annotation(
    x=5.2, 
    y=59.0,
    xref="x", 
    yref="y",
    text=stats_text,
    showarrow=True,
    arrowhead=2,
    arrowsize=1,
    arrowwidth=1,
    arrowcolor="#666666",
    ax=40,
    ay=-40,
    font=dict(size=12, color="#1e293b"),
    align="left",
    bgcolor="rgba(255, 255, 255, 0.92)",
    bordercolor="#cbd5e1",
    borderwidth=1,
    borderpad=8
)

# 7. Layout adjustments 
fig.update_layout(
    title='<b>Relating Opening Depth, Game Duration, and Popularity Across Top 8 Families</b>',
    xaxis_title='Average Opening Depth (Ply)',
    yaxis_title='Average Total Game Duration (Turns)',
    width=950,
    height=600,
    font=dict(size=13, family="Arial"),
    title_font=dict(size=16),
    margin=dict(l=60, r=40, t=80, b=60)
)

fig.show()
```

**Interpretation of the Graph**

Weak linear relationship; theoretical opening depth does not dictate total match duration. Instead, game length is driven by the pawn structure and tactical nature of the specific opening family—as demonstrated by the contrast between the Ruy Lopez (long games) and Italian Game (short games) despite both sharing similar theory depths.

## 2.2. Bar Graph

### **EDA 2 :Is game duration primarily driven by opening selection or by player skill level (avg_rating / skill_tier)?**

*In order to conduct analysis, I have grouped bar chart to compare average game durations (in turns) across the top 8 opening families,which is segmented further by player skill tier (Beginner, Intermediate, and Advanced/Pro).*

**Details of the code**

1. **Rating Tier Segmentation**
Calculates the average match rating (avg_rating) from white and black player ratings and maps each game into a distinct skill_tier (Beginner, Intermediate, or Advanced/Pro) using a custom classification function.

2. **Data Filtering**
Isolates the dataset to focus exclusively on the top 6 most frequent chess opening families.

3. **Data Aggregation**
Groups the data by both opening_family and skill_tier to compute the average game duration (mean_turns) and sample count (game_count) for each combination.

4. **Global Metric Computation**
Calculates overall dataset metrics (total sample size and global mean turns) to establish a baseline for annotations.

5. **Grouped Bar Chart Construction**
Builds a grouped bar chart using Plotly Express (px.bar), displaying opening families on the X-axis, average turns on the Y-axis, and splitting/coloring bars by skill tier.

6. **Trace & Label Configuration**
Formats the data labels to automatically show the exact turn values on top of each bar.

7. **Summary Annotation & Layout Styling**
Injects a formatted Quantitative Summary Box into the layout, applies a clean white template, and customizes dimensions, fonts, margins, and the horizontal legend position.

```python
import plotly.express as px
import pandas as pd

# 1. Define rating tier segmentation
chess['avg_rating'] = (chess['white_rating'] + chess['black_rating']) / 2

def get_skill_tier(rating):
    if rating < 1400:
        return 'Beginner (<1400)'
    elif rating < 1800:
        return 'Intermediate (1400-1800)'
    else:
        return 'Advanced / Pro (1800+)'

chess['skill_tier'] = chess['avg_rating'].apply(get_skill_tier)

# 2. Filter for top opening families
top_families = chess['opening_family'].value_counts().head(6).index
filtered_chess = chess[chess['opening_family'].isin(top_families)]

# 3. Group by opening family and skill tier to calculate average turns
bar_stats = filtered_chess.groupby(['opening_family', 'skill_tier']).agg(
    mean_turns=('turns', 'mean'),
    game_count=('turns', 'count')
).reset_index()
    
# 4. Compute overall dataset metrics for the summary box
overall_mean_turns = filtered_chess['turns'].mean()
total_games = len(filtered_chess)

# 5. Create grouped bar chart
fig = px.bar(
    bar_stats,
    x='opening_family',
    y='mean_turns',
    color='skill_tier',
    barmode='group',
    text_auto='.1f',
    hover_data={'game_count': True, 'mean_turns': ':.2f'},
    title='<b>Average Game Duration by Opening Family & Skill Tier</b>',
    labels={
        'opening_family': 'Opening Family',
        'mean_turns': 'Average Game Duration (Turns)',
        'skill_tier': 'Skill Tier',
        'game_count': 'Total Games'
    },
    color_discrete_sequence=['#38bdf8', '#818cf8', '#c084fc']
)

# 6. Styling and layout adjustments
fig.update_traces(
    textfont_size=10, 
    textangle=0, 
    textposition="outside", 
    cliponaxis=False
)

# 7. Add Quantitative Summary Box
stats_text = (
    f"<b>Quantitative Summary</b><br>"
    f"• Dataset Mean Turns: <b>{overall_mean_turns:.1f}</b><br>"
    f"• Total Sample Size (n): <b>{total_games:,}</b>"
)

fig.add_annotation(
    x=0.98, 
    y=0.96,
    xref="paper", 
    yref="paper",
    text=stats_text,
    showarrow=False,
    font=dict(size=11, color="#1e293b"),
    align="left",
    bgcolor="rgba(255, 255, 255, 0.92)",
    bordercolor="#cbd5e1",
    borderwidth=1,
    borderpad=8
)

fig.update_layout(
    template="plotly_white",
    width=1050,
    height=600,
    font=dict(size=13, family="Arial"),
    title_font=dict(size=16),
    legend=dict(title="<b>Skill Tiers</b>", orientation="h", y=1.05, x=0.01),
    margin=dict(l=60, r=40, t=100, b=80),
    xaxis_tickangle=-25
)

fig.show()
```

**Interpretation of Graph:**

Beginner games are significantly shorter due to early blunder-induced resignations, while advanced games are systematically longer due to positional defense and endgame maneuvering. Hence, player rating level is a far stronger predictor of game duration than the opening choice itself.


# Part 3 : Streamlit app

*The code builds a complete, interactive Streamlit web application designed to explore the chess dataset, analyzing how opening choices, theory depth, and player skill tiers influence game duration.*

**Core Components of the Application**

1. **Page Setup & Caching (`@st.cache_data`):** Configures a wide-layout Streamlit dashboard titled *"Chess Opening & Duration Explorer"* and efficiently loads and caches `chess_cleaned.csv`. It validates required columns and computes derived variables like `avg_rating` and `skill_tier` (Beginner, Intermediate, Advanced).
2. **Interactive Sidebar Controls:** Provides a suite of real-time filters:
3. **Rated Status** (multiselect)
4. **Victory Status** (multiselect)
5. **Opening Families** (multiselect for top 8 families)
6. **Skill Tier** (dropdown selector)


* **Top Metric Cards:** Dynamically computes and displays four key quantitative metrics based on the filtered data:
1. *Total Games in View*
2. *Mean Turns*
3. *Opening Depth Correlation ($r$)* via SciPy's `stats.pearsonr`
4. *Pearson $p$-value*


* **Visualization 1 (Altair Bubble Chart):** Plots average opening depth (`opening_ply`) against average game duration (`turns`), using bubble size to represent total games played and color-coding by opening family with embedded text labels.
* **Visualization 2 (Faceted Bar Chart):** Creates small multiples using Altair column faceting to compare average game durations across opening families broken down by player skill tier, featuring independent X-axes and rotated labels.
* **Data Preview Table:** Renders an interactive, scrollable preview table (`st.dataframe`) showing the top 100 rows of the currently filtered dataset.

```python
import streamlit as st
import pandas as pd
import altair as alt
import scipy.stats as stats

# -----------------------------------------------------------------------------
# PAGE CONFIGURATION
# -----------------------------------------------------------------------------
st.set_page_config(
    page_title="Chess Opening & Duration Explorer",
    page_icon="♟️",
    layout="wide"
)

# -----------------------------------------------------------------------------
# DATA LOADING & PREPROCESSING (Cached)
# -----------------------------------------------------------------------------
@st.cache_data
def load_data():
    df = pd.read_csv('chess_cleaned.csv')
    
    # columns 
    required_cols = ['id', 'opening_family', 'turns', 'opening_ply', 'rated', 'victory_status', 'white_rating', 'black_rating']
    for col in required_cols:
        if col not in df.columns:
            st.error(f"Missing required column: '{col}' in chess_cleaned.csv.")
            st.stop()
            
    # Calculate average rating and assign skill tiers
    df['avg_rating'] = (df['white_rating'] + df['black_rating']) / 2
    
    def get_skill_tier(rating):
        if rating < 1400:
            return 'Beginner (< 1400)'
        elif rating <= 1800:
            return 'Intermediate (1400 - 1800)'
        else:
            return 'Advanced (> 1800)'

    df['skill_tier'] = df['avg_rating'].apply(get_skill_tier)
    return df

try:
    chess = load_data()
except Exception as e:
    st.error(f"Error loading data: {e}")
    st.stop()

# -----------------------------------------------------------------------------
# SIDEBAR CONTROLS (All 3 Filters Included)
# -----------------------------------------------------------------------------
st.sidebar.header("🎛️ Dashboard Filters")

# 1. Rated Status Filter
rated_options = chess['rated'].unique().tolist()
selected_rated = st.sidebar.multiselect(
    "Select Rated Status",
    options=rated_options,
    default=rated_options
)

# 2. Victory Status Filter
victory_options = chess['victory_status'].unique().tolist()
selected_victory = st.sidebar.multiselect(
    "Select Victory Status",
    options=victory_options,
    default=victory_options
)

# 3. Opening Families Multiselect Filter
top_families_list = chess['opening_family'].value_counts().head(8).index.tolist()
selected_families = st.sidebar.multiselect(
    "Select Opening Families:",
    options=top_families_list,
    default=top_families_list
)

# 4. Skill Tier Filter
rating_tier = st.sidebar.selectbox(
    "Select Skill Tier",
    options=["All Tiers", "Beginner (< 1400)", "Intermediate (1400 - 1800)", "Advanced (> 1800)"]
)

# Apply All Sidebar Filters
filtered_df = chess.copy()
if selected_rated:
    filtered_df = filtered_df[filtered_df['rated'].isin(selected_rated)]
if selected_victory:
    filtered_df = filtered_df[filtered_df['victory_status'].isin(selected_victory)]
if selected_families:
    filtered_df = filtered_df[filtered_df['opening_family'].isin(selected_families)]

if rating_tier == "Beginner (< 1400)":
    filtered_df = filtered_df[filtered_df['skill_tier'] == 'Beginner (< 1400)']
elif rating_tier == "Intermediate (1400 - 1800)":
    filtered_df = filtered_df[filtered_df['skill_tier'] == 'Intermediate (1400 - 1800)']
elif rating_tier == "Advanced (> 1800)":
    filtered_df = filtered_df[filtered_df['skill_tier'] == 'Advanced (> 1800)']

# -----------------------------------------------------------------------------
# MAIN DASHBOARD HEADER & METRICS
# -----------------------------------------------------------------------------
st.title("♟️ Chess Opening & Duration Explorer")
st.markdown("Explore the relationship between chess opening families, player skill tiers, opening theory depth, and game duration.")

st.markdown("---")

col1, col2, col3, col4 = st.columns(4)

total_games = len(filtered_df)
mean_turns = filtered_df['turns'].mean() if total_games > 0 else 0

if 'opening_ply' in filtered_df.columns and total_games > 1:
    r_val, p_val = stats.pearsonr(filtered_df['opening_ply'], filtered_df['turns'])
else:
    r_val, p_val = 0.0, 1.0

col1.metric("Total Games in View", f"{total_games:,}")
col2.metric("Mean Turns", f"{mean_turns:.1f} turns")
col3.metric("Opening Depth Correlation (r)", f"{r_val:.3f}")
col4.metric("Pearson p-value", "p < 0.001" if p_val < 0.001 else f"p = {p_val:.3f}")

st.markdown("---")

# -----------------------------------------------------------------------------
# VISUALIZATION 1: ALTAIR BUBBLE CHART (Opening Depth vs Game Duration)
# -----------------------------------------------------------------------------
st.subheader("📊 Opening Depth vs. Game Duration & Popularity")
st.markdown("Examining how opening theory depth (`opening_ply`) relates to overall game duration (`turns`) across selected opening families.")

if not filtered_df.empty:
    family_rel = filtered_df.groupby('opening_family').agg(
        mean_ply=('opening_ply', 'mean'),
        mean_turns=('turns', 'mean'),
        game_count=('turns', 'count')
    ).reset_index()

    bubble_chart = alt.Chart(family_rel).mark_circle(opacity=0.85, stroke='DarkSlateGrey', strokeWidth=1).encode(
        x=alt.X('mean_ply:Q', title='Average Opening Depth (Ply)'),
        y=alt.Y('mean_turns:Q', title='Average Total Game Duration (Turns)', scale=alt.Scale(zero=False)),
        size=alt.Size('game_count:Q', title='Total Games Played', legend=None, scale=alt.Scale(range=[100, 1000])),
        color=alt.Color('opening_family:N', title='Opening Family', legend=None),
        tooltip=['opening_family', 'mean_ply', 'mean_turns', 'game_count']
    ).properties(
        width=900,
        height=500,
        title='Relating Opening Depth, Game Duration, and Popularity Across Selected Families'
    )

    text_labels = alt.Chart(family_rel).mark_text(align='center', baseline='bottom', dy=-12, fontSize=11).encode(
        x='mean_ply:Q',
        y='mean_turns:Q',
        text='opening_family:N'
    )

    st.altair_chart(bubble_chart + text_labels, use_container_width=True)
else:
    st.warning("No data available with the current filter selection.")

st.markdown("---")

# -----------------------------------------------------------------------------
# VISUALIZATION 2: FACETED BAR CHART BY SKILL TIER (All Names Displayed)
# -----------------------------------------------------------------------------
st.subheader("📈 Average Game Duration by Opening Family & Skill Tier")
st.markdown("Ranking opening families broken down across player skill tiers using faceted small multiples.")

if len(filtered_df) > 0:
    bar_stats = filtered_df.groupby(['opening_family', 'skill_tier']).agg(
        mean_turns=('turns', 'mean'),
        game_count=('turns', 'count')
    ).reset_index()

    bar_chart = alt.Chart(bar_stats).mark_bar().encode(
        x=alt.X(
            'opening_family:N', 
            title='Opening Family', 
            axis=alt.Axis(
                labelAngle=-90,         
                labelOverlap=False       
            )
        ),
        y=alt.Y('mean_turns:Q', title='Mean Turns'),
        color=alt.Color('skill_tier:N', title='Skill Tier'),
        column=alt.Column('skill_tier:N', title='skill_tier'),
        tooltip=['opening_family', 'skill_tier', alt.Tooltip('mean_turns:Q', format='.1f'), 'game_count']
    ).properties(
        width=200,
        height=400
    ).resolve_axis(
        x='independent'
    )
    
    st.altair_chart(bar_chart, use_container_width=True)
else:
    st.warning("Please select at least one opening family / skill tier combination to display the bar chart.")

st.markdown("---")

# -----------------------------------------------------------------------------
# DATASET PREVIEW
# -----------------------------------------------------------------------------
st.subheader("📋 Filtered Dataset Preview")
preview_cols = [col for col in ['id', 'rated', 'turns', 'victory_status', 'opening_family', 'skill_tier', 'opening_ply', 'avg_rating'] if col in filtered_df.columns]
st.dataframe(filtered_df[preview_cols].head(100), use_container_width=True)
```
